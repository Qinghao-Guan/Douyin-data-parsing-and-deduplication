```python
import pandas as pd

# File paths
input_file = "zeeschuimer-export-douyin.com-2026-09-30T153907.ndjson"
output_file = "douyin_deduplicated.xlsx"

# Read the NDJSON file as a DataFrame
df = pd.read_json(input_file, lines=True)
data_flat = pd.json_normalize(
    df["data"],
    sep="_"
)

# Add prefix to clearly identify fields originating from "data"
data_flat = data_flat.add_prefix("data_")

# Drop original nested data column
df = df.drop(columns=["data"])

# Combine original columns with flattened data columns
df = pd.concat(
    [
        df.reset_index(drop=True),
        data_flat.reset_index(drop=True)
    ],
    axis=1
)

print(f"\nColumns after flattening: {len(df.columns):,}")
print(df.columns.tolist())
```

```python
df
```

```python
# Remove duplicates based on caption
df_clean = df.drop_duplicates(
    subset=["data_desc"],
    keep="first"
)

# Reset row index
df_clean = df_clean.reset_index(drop=True)

print(f"Rows after deduplication: {len(df_clean):,}")
print(f"Duplicates removed: {len(df) - len(df_clean):,}")

# Convert nested objects to strings so Excel can store them
for col in df_clean.columns:
    if df_clean[col].apply(lambda x: isinstance(x, (dict, list))).any():
        df_clean[col] = df_clean[col].astype(str)

# Save to Excel
df_clean.to_excel(
    output_file,
    index=False,
    engine="openpyxl"
)

print(f"Saved to: {output_file}")
```

**Add a new column for the video's URL link**

```python
# --------------------------------------------------
input_file = "/content/douyin_deduplicated excel.xlsx"
output_file = "/content/douyin_deduplicated_with_video_url.xlsx"


# Read Excel file
df = pd.read_excel(input_file)

print(f"Number of rows: {len(df):,}")
print(f"Number of columns: {len(df.columns):,}")


# Create a clean Douyin video URL

df["video_url"] = df["data_aweme_id"].apply(
    lambda x: f"https://www.douyin.com/video/{int(x)}"
    if pd.notna(x)
    else None
)

# Move video_url near the front

columns = df.columns.tolist()

columns.remove("video_url")

# Put video_url after caption, if caption exists
if "caption" in columns:
    caption_index = columns.index("caption")
    columns.insert(caption_index + 1, "video_url")
else:
    columns.insert(0, "video_url")

df = df[columns]

# Save
df.to_excel(
    output_file,
    index=False,
    engine="openpyxl"
)

print(f"Saved to: {output_file}")
```
